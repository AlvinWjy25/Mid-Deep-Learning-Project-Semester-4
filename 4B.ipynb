```python
import numpy as np
import pandas as pd
```

1. Define Variable & Activation Function

```python
def relu(x):
    return np.maximum(0, x)

def relu_deriv(x):
    return (x > 0).astype(float)

def mse_loss(y_pred, y_true):
    return 0.5 * (y_pred - y_true)**2
```

```python
lr = 0.01
b = 1
```

```python
x = np.array([
    [-4, 3],
    [1, 0]
])

y = np.array([
    [-1],
    [5]
])

W1 = np.array([
    [2.0, -1.0, 1.0], 
    [1.0,  -2.0, -1.0]
])
b1 = np.full((1, 3), b)

W2 = np.array([
    [1.0,  3.0], 
    [-2.0, 1.0], 
    [2.0, -2.0]
])
b2 = np.full((1, 2), b)


W3 = np.array([
    [-2.0],
    [3.0]
])
b3 = np.full((1, 1), b)
```

2. Hidden Layer 1 & 2

```python
Z1 = np.dot(x, W1) + b1
A1 = relu(Z1)

Z2 = np.dot(A1, W2) + b2
A2 = relu(Z2)

Z3 = np.dot(A2, W3) + b3
output = Z3 

n = x.shape[0]
loss_mse = np.mean(np.square(y - output))
```

```python
print("=" * 20, "Hasil", "=" * 20)
print(f"Z1 = \n {Z1}")
print(f"Z2 = \n {Z2}")
print(f"Z3 = \n {Z3}")

print(f"A1 = \n {A1}")
print(f"A2 = \n {A2}")

print("=" * 47)
print(f"Prediksi:\n{output}")
print(f"MSE Loss:\n{loss_mse}")
```

3. Backward Pass

```python
delta_loss_out = (2 / n) * (output - y)

deriv_W3 = np.dot(A2.T, delta_loss_out)
deriv_B3 = np.sum(delta_loss_out, axis = 0, keepdims = True)

deriv_A2 = np.dot(delta_loss_out, W3.T)
deriv_Z2 = deriv_A2 * relu_deriv(Z2)
deriv_W2 = np.dot(A1.T, deriv_Z2)
deriv_B2 = np.sum(deriv_Z2, axis=0, keepdims=True)


deriv_A1 = np.dot(deriv_A2, W2.T)
deriv_Z1 = deriv_A1 * relu_deriv(Z1)
deriv_W1 = np.dot(x.T, deriv_Z1)
deriv_B1 = np.sum(deriv_Z1, axis=0, keepdims=True)
```

```python
print("=" * 20, "Backward Pass", "=" * 20)

print("derivative W3:\n", deriv_W3)
print("derivative B3:\n", deriv_B3)

print("derivative A2:\n", deriv_A2)
print("derivative Z2:\n", deriv_Z2)
print("derivative W2:\n", deriv_W2)
print("derivative B2:\n", deriv_B2)

print("derivative A1:\n", deriv_A1)
print("derivative Z1:\n", deriv_Z1)
print("derivative W1:\n", deriv_W1)
print("derivative B1:\n", deriv_B1)
```

4. Update Weights

```python
W1_new = W1 - lr * deriv_W1
W2_new = W2 - lr * deriv_W2
W3_new = W3 - lr * deriv_W3

B1_new = b1 - lr * deriv_B1
B2_new = b2 - lr * deriv_B2
B3_new = b3 - lr * deriv_B3
```

```python
print("=" * 20 ,"\nWeights Baru (setelah 1 step):", "=" * 20)
print("W1 New:\n", W1_new)
print("W2 New:\n", W2_new)
print("W3 New:\n", W3_new, '\n\n')

print("B1 New:\n", B1_new)
print("B2 New:\n", B2_new)
print("B3 New:\n", B3_new)
```
